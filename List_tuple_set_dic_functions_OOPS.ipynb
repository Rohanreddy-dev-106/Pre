```python
#*args **kwargs
def multiply(*args):
    product=1
    for i in args:
        product*=i;
    return product

def display(**kwargs):
    print(type(kwargs),kwargs)
    for (key,value) in kwargs.items():
        print(key , value)
```

```python
print(multiply(1,2,3))
display(name='rohan',age=21)
```
