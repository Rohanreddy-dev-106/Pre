```python
#superpowe methods
#in Python, __add__ is a magic method that defines how the + operator works for objects.
class Fraction:
    def __init__(self,num,den):
        self.num=num
        self.den=den
    def __str__(self):
        return f"{self.num}/{self.den}"
    def __add__(self,other):
        n = self.num * other.den + other.num * self.den
        d=self.den*other.den
        return f"{n}/{d}"
    def __sub__(self,other):
        n = self.num * other.den - other.num * self.den
        d=self.den*other.den
        return f"{n}/{d}"
    def __mul__(self,other):
        n = self.num *  other.num
        d=self.den*other.den
        return f"{n}/{d}"

```

```python
f1=Fraction(5,2)
f2=Fraction(1,2)
print(f1)
print(f1+f2)
print(f1-f2)
print(f1*f2)
```

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
```

```python
class LinkedList:
    def createnoders(self):
        data=int(input("enter data"))
        head=None
        tail=None
        while data !=-1:
             node=Node(data)
             if head==None:
                head=node
                tail=node
             else:
                 tail.next=node
                 tail=tail.next
             data=int(input("enter data"))
        return head

    def linkedListPrint(self,head):
        while head!=None:
            print("{}->".format(head.data),end="")
            head=head.next
        print("None",end="")

    def insertList(self, data, index, head):
         i = 0
         temp = head
         if index==0:
            new_node=Node(data)
            new_node.next=temp
            return new_node

         while i != index - 1:
            temp = temp.next
            i += 1

         new_node = Node(data)
         nextNode = temp.next
         temp.next = new_node
         new_node.next = nextNode

         return head
```

```python
L=LinkedList()
h=L.createnoders()
L.linkedListPrint(h)
x=L.insertList(106,2,h)
print("")
L.linkedListPrint(x)
```

```python
#Inheritance & Polymorphism
class User:
    counter=0
    def __init__(self,name,age):
        self.name=name
        self.age=age
        User.counter=User.counter+1
    def getName(self):
        return self.name
    def getAge(self):
        return self.age
    @staticmethod
    def getCount():
        return User.counter


```

```python
class Student(User):
    def __init__(self, name, age):
        super().__init__(name, age) #super constructor
        super().getName()
s1=Student("Rohan",22)
s2=Student("Sai",23)
s3=Student("Sai",23)
print(s2.getName())
print(s2.getAge())
s2.counter

```
