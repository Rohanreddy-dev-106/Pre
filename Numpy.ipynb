```python
store = pd.read_csv('/content/store.csv', encoding='latin1')
map={}
for col in store.columns:
    map[col]=coloums_checker(store[col])


```

```python
store.isnull().mean() < 0.6
store=store.loc[:,store.isnull().mean() < 0.6] #Keep all rows, but only keep columns where missing values are less than 60%.
```

```python
filtered_map={}
for key,value in map.items():
        if value not in filtered_map:#if not exist create else append it
           filtered_map[value]=[]#create a list
        filtered_map[value].append(key)#append current value
filtered_map

```

```python
#numerical_semmary:
numeric_summary={}
for col,values in filtered_map.items():
    if col =="numeric":
        for val in values:
            if val not in numeric_summary:
                numeric_summary[val]={}
            numeric_summary[val]["count"] = store[val].count()
            numeric_summary[val]["mean"] = store[val].mean()
            numeric_summary[val]["median"] = store[val].median()
            numeric_summary[val]["std"] = store[val].std()
            numeric_summary[val]["min"] = store[val].min()
            numeric_summary[val]["max"] = store[val].max()
numeric_summary
```

```python
#categorical_summary
categorical_summary = {}

for col, values in filtered_map.items():
    if col == "categorical":
        for val in values:
            if val not in categorical_summary:
                categorical_summary[val] = {}

            categorical_summary[val]["count"] = store[val].count()
            categorical_summary[val]["unique"] = store[val].nunique()
            categorical_summary[val]["freq"] = store[val].value_counts().iloc[0]
            categorical_summary[val]["most_frequent"] = store[val].mode().iloc[0]
            categorical_summary[val]["most_frequent"] = store[val].mode().iloc[0]



categorical_summary
```

```python
datetime_summary = {}

for col, values in filtered_map.items():
    if col == "datetime":
        for val in values:
            if val not in datetime_summary:
                datetime_summary[val] = {}

            datetime_summary[val]["count"] = store[val].count()
            datetime_summary[val]["start"] = store[val].min()
            datetime_summary[val]["end"] = store[val].max()
datetime_summary
```

```python
missing_summary = {}

for col in store.columns:
    if col not in missing_summary:
        missing_summary[col] = {}

    missing_summary[col]["missing_count"] = store[col].isnull().sum()
    missing_summary[col]["missing_percent"] = round(store[col].isnull().mean() * 100, 2)

missing_summary
```

```python

def coloums_checker(col):
    if pd.api.types.is_numeric_dtype(col):
        return "numeric"
    elif pd.api.types.is_string_dtype(col):
        return "categorical"

    elif pd.api.types.is_datetime64_any_dtype(col):
        return "datetime"

    elif pd.api.types.is_bool_dtype(col):
        return "boolean"

    else:
        return "unknown"
```
